```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("D:\Multi Disease Model\Data\Breast_Cancer_Data.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Breast_Cancer')
plt.title('Breast Cancer Distribution')
plt.xlabel('Breast Cancer')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Age', kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='Breast_Cancer', y='BMI')
plt.title('BMI vs Breast Cancer')
plt.xlabel('Breast Cancer')
plt.ylabel('BMI')
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='Breast_Cancer', y='Tumor_Size_mm')
plt.title('Tumor Size vs Breast Cancer')
plt.xlabel('Breast Cancer')
plt.ylabel('Tumor Size (mm)')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='Age',
    y='Tumor_Size_mm',
    hue='Breast_Cancer'
)
plt.title('Age vs Tumor Size')
plt.xlabel('Age')
plt.ylabel('Tumor Size (mm)')
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(
    data=df,
    x='Family_History',
    hue='Breast_Cancer'
)
plt.title('Family History vs Breast Cancer')
plt.xlabel('Family History')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(
    data=df,
    x='Lump_Present',
    hue='Breast_Cancer'
)
plt.title('Lump Presence vs Breast Cancer')
plt.xlabel('Lump Present')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10, 7))

numeric_df = df.select_dtypes(include=np.number)

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```
