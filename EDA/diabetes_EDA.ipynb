```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("D:/Multi Disease Model/Data/Diabetes_Data.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Diabetes')
plt.title('Diabetes Distribution')
plt.xlabel('Diabetes')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Glucose', kde=True)
plt.title('Glucose Distribution')
plt.xlabel('Glucose')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Age', kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='Diabetes', y='BMI')
plt.title('BMI vs Diabetes')
plt.xlabel('Diabetes')
plt.ylabel('BMI')
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='Diabetes', y='Glucose')
plt.title('Glucose vs Diabetes')
plt.xlabel('Diabetes')
plt.ylabel('Glucose')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='Glucose', hue='Diabetes')
plt.title('Age vs Glucose')
plt.xlabel('Age')
plt.ylabel('Glucose')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Glucose', y='HbA1c', hue='Diabetes')
plt.title('Glucose vs HbA1c')
plt.xlabel('Glucose')
plt.ylabel('HbA1c')
plt.show()
```

```python
plt.figure(figsize=(10, 7))

numeric_df = df.select_dtypes(include=np.number)

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```
